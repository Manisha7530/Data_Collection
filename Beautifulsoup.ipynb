```python
from bs4 import BeautifulSoup
```

```python
with open("htmls/page1.html") as f:
    contant = f.read()
```

```python
soup = BeautifulSoup(contant, "html.parser")
```

```python
h3s = soup.find_all("h3")
```

```python
h3s
```

```python
for h3 in h3s:
    print(h3.find("a")["title"])
```

```python
articles = soup.select("article.product_pod")
```

```python
articles
```

```python
for article in articles:
    title = article.find("h3").find("a")["title"]
    print(title)
```

```python
items = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text
    items.append([title, price])
```

```python
items
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(items, columns=["Books titles", "Price"])
```

```python
df
```

```python
items = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    price = article.select_one("p.price_color").text
    rating_element = article.select_one("p.star-rating")
    rating = rating_element['class'][1]
    items.append([title, price, rating])
```

```python
items
```

```python
df1 = pd.DataFrame(items, columns=['Books titel', 'Prices', 'Rating'])
```

```python
df1
```

```python
df1.to_csv("HTML-Books.csv", index=False)
```
